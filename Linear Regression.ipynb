Let's split the data into training and testing sets for better results

```python
from sklearn.model_selection import  train_test_split
```

```python
df.columns
```

```python
x = df[[ "engine-size", "horsepower", "highway-mpg"]]
```

```python
y = df["price"]
```

```python
x_train, x_test, y_train, y_test = train_test_split(x, y, test_size = 0.3, random_state = 101)
```

# Training the model

```python
from sklearn.linear_model import LinearRegression
```

```python
lm = LinearRegression()
```

```python
lm.fit(x_train, y_train)
```

```python
lm.intercept_
```

```python
lm.coef_
```

```python
predictions = lm.predict(x_test)
```

```python
plt .scatter(y_test, predictions)
plt.xlabel("Y Test(True values)")
plt.ylabel("Predicted_values")
```

```python
from sklearn import metrics
```

```python
print('MAE', metrics.mean_absolute_error(y_test,predictions))
print('MSE', metrics.mean_squared_error(y_test,predictions))
print('RMSE',np.sqrt(metrics.mean_squared_error(y_test,predictions)))
print("Rsquared", lm.score())
```

```python
metrics.explained_variance_score(y_test, predictions)
```

Plot a histogram of the residuals and make sure it looks normally distributed. Use either seaborn displot, or just plt.hist().

```python
sns.displot(y_test - predictions, bins = 50, kde = True)
```
